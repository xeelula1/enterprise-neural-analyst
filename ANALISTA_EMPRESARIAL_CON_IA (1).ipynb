# Módulo 1: Ingesta de Documentos, Chunking y Embeddings.

```python
# 2. Definimos un texto bíblico de prueba y aplicamos Chunking
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Texto bíblico de prueba
texto_biblico = """
SALMO 91: EL REFUGIO DEL CREYENTE

1 El que habita al abrigo del Altísimo morará bajo la sombra del Omnipotente.
2 Diré yo a Jehová: Esperanza mía, y castillo mío; Mi Dios, en quien confiaré.
3 Él te librará del lazo del cazador, de la peste destructora.
4 Con sus plumas te cubrirá, y debajo de sus alas estarás seguro; escudo y adarga es su verdad.
5 No temerás al terror nocturno, ni saeta que vuele de día,
6 Ni pestilencia que habite en oscuridad, ni mortandad que en medio del día destruya.

PROVERBIOS 3: SABIDURÍA Y CONFIANZA

1 Hijo mío, no te olvides de mi ley, y tu corazón guarde mis mandamientos;
2 Porque abundancia de días y años de vida y paz te aumentarán.
3 Nunca se mepartan de ti la misericordia y la verdad; átalas a tu cuello, escríbelas en la tabla de tu corazón.
5 Fíate de Jehová de todo tu corazón, y no te apoyes en tu propia prudencia.
6 Reconócelo en todos tus caminos, y él enderezará tus veredas.
"""

# Configuración del Splitter
# chunk_size: máximo de caracteres por fragmento
# chunk_overlap: caracteres de solapamiento para mantener el contexto entre versículos
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    length_function=len
)

# Generamos los fragmentos (chunks)
chunks = text_splitter.split_text(texto_biblico)

print(f"✅ Texto bíblico procesado exitosamente.")
print(f"🔹 Total de Chunks creados: {len(chunks)}\n")

# Mostramos los chunks generados
for i, chunk in enumerate(chunks):
    print(f"--- CHUNK {i+1} ---")
    print(chunk)
    print()
```

```python
# 3. Guardamos los chunks en ChromaDB con Embeddings
import chromadb
from sentence_transformers import SentenceTransformer

# Cargamos el modelo de Embeddings (multilingüe y ligero, ideal para español)
print("⏳ Cargando modelo de embeddings (SentenceTransformers)...")
embedding_model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# Inicializamos la base de datos ChromaDB (en memoria para pruebas en Colab)
chroma_client = chromadb.Client()

# Creamos una colección llamada 'johan_knowledge'
collection = chroma_client.create_collection(name="johan_knowledge")

# Convertimos cada chunk en embedding y lo insertamos en ChromaDB
for i, chunk in enumerate(chunks):
    vector = embedding_model.encode(chunk).tolist()
    collection.add(
        documents=[chunk],
        embeddings=[vector],
        metadatas=[{"fuente": "Texto_Biblico_Prueba", "chunk_id": i}],
        ids=[f"id_{i}"]
    )

print(f"✅ ¡Éxito! Johan ha almacenado {len(chunks)} vectores en ChromaDB.")
```

# PROBEMOS A JOHAN

```python
# 4. Hacemos una consulta semántica a la base de datos de Johan
pregunta_usuario = "¿Qué dice sobre dónde buscar refugio o protección?"

# Convertimos la pregunta del usuario en un vector numérico
vector_pregunta = embedding_model.encode(pregunta_usuario).tolist()

# Buscamos en ChromaDB los 2 fragmentos más parecidos semánticamente
resultados = collection.query(
    query_embeddings=[vector_pregunta],
    n_results=2
)

print(f"🔍 Pregunta realizada a Johan: '{pregunta_usuario}'\n")
print("📌 Contexto exacto recuperado por Johan de ChromaDB:\n")

for i, doc in enumerate(resultados['documents'][0]):
    metadata = resultados['metadatas'][0][i]
    print(f"--- Resultado {i+1} (Fuente: {metadata['fuente']}) ---")
    print(doc)
    print()
```

# Módulo 2 (Conexión con el LLM y Generación RAG)

```python
# 1. Instalamos transformers y accelerate para correr el LLM en Colab
!pip install -q transformers accelerate bitsandbytes
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

print("⏳ Cargando Llama 3 (Versión Pública Cuantizada)...")

# Modelo de Llama 3 de acceso libre y optimizado para consumo de memoria
model_id = "unsloth/llama-3-8b-Instruct-bnb-4bit"

generator = pipeline(
    "text-generation",
    model=model_id,
    device_map="auto"
)

def preguntar_a_johan(pregunta: str, top_k: int = 2):
    # 1. Recuperación de fragmentos desde ChromaDB
    vector_pregunta = embedding_model.encode(pregunta).tolist()
    resultados = collection.query(
        query_embeddings=[vector_pregunta],
        n_results=top_k
    )

    contexto = "\n---\n".join(resultados['documents'][0])

    # 2. Prompt estructurado en formato Llama 3
    system_prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>
Eres Johan, un analista de Inteligencia Artificial empresarial.
Responde a la consulta del usuario utilizando ÚNICAMENTE la información presente en el contexto adjunto.
Sé claro, profesional, directo y responde con total fluidez en español. Si la información no está en el contexto, indica sencillamente que no dispones de ella.<|eot_id|>
<|start_header_id|>user<|end_header_id|>
CONTEXTO DISPONIBLE:
{contexto}

PREGUNTA DEL USUARIO:
{pregunta}<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>"""

    # 3. Generación con Llama 3
    respuesta = generator(
        system_prompt,
        max_new_tokens=250,
        temperature=0.3,
        do_sample=True
    )

    # Extraer únicamente la respuesta de Johan
    texto_completo = respuesta[0]['generated_text']
    respuesta_johan = texto_completo.split("<|start_header_id|>assistant<|end_header_id|>")[-1].replace("<|eot_id|>", "").strip()

    print(f"🤖 **JOHAN (Llama 3):**\n{respuesta_johan}\n")
    print("📌 **Contexto exacto recuperado de ChromaDB:**")
    print(contexto)

print("✅ ¡Johan configurado con Llama 3 al 100%!")
```

```python
# Prueba final del Módulo 2: Johan respondiendo con Llama 3
preguntar_a_johan("¿Qué protección se promete contra el terror de la noche y las enfermedades?")
```
